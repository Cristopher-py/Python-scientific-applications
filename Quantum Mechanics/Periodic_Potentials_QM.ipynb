### Schrödinger Equation with Bloch's Theorem

We begin with the time-independent Schrödinger equation in one dimension:

$$
\left[-\frac{\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)\right]\psi(x) = E\psi(x)
$$

where:
- $\hbar$ is the reduced Planck constant
- $m$ is the particle mass
- $V(x)$ is the potential energy (periodic with period $L$)
- $E$ is the energy eigenvalue
- $\psi(x)$ is the wave function

For a strictly periodic potential $V(x)$ with period $L$:

$$
V(x + L) = V(x) \quad \forall x \in \mathbb{R}
$$

Bloch's theorem states that the wave functions can be written in the form:

$$
\psi_k(x) = e^{ikx}u_k(x)
$$

where:
- $k$ is the wave vector (real number)
- $u_k(x)$ is a periodic function with period $L$: $u_k(x + L) = u_k(x)$
- $k$ is restricted to the first Brillouin zone: $k \in [-\pi/L, \pi/L]$

Compute the derivatives:

$$
\frac{d\psi_k(x)}{dx} = \frac{d}{dx}\left[e^{ikx}u_k(x)\right] = e^{ikx}\left[iku_k(x) + \frac{du_k(x)}{dx}\right]
$$

$$
\frac{d^2\psi_k(x)}{dx^2} = \frac{d}{dx}\left[e^{ikx}\left(iku_k(x) + \frac{du_k(x)}{dx}\right)\right] = e^{ikx}\left[-k^2u_k(x) + 2ik\frac{du_k(x)}{dx} + \frac{d^2u_k(x)}{dx^2}\right]
$$

Substituting into the Schrödinger equation:

$$
-\frac{\hbar^2}{2m}e^{ikx}\left[-k^2u_k(x) + 2ik\frac{du_k(x)}{dx} + \frac{d^2u_k(x)}{dx^2}\right] + V(x)e^{ikx}u_k(x) = E_ke^{ikx}u_k(x)
$$

Dividing both sides by $e^{ikx}$ (nonzero for all $x$):

$$
-\frac{\hbar^2}{2m}\left[\frac{d^2u_k(x)}{dx^2} + 2ik\frac{du_k(x)}{dx} - k^2u_k(x)\right] + V(x)u_k(x) = E_ku_k(x)
$$

Expanding and rearranging terms:

$$
-\frac{\hbar^2}{2m}\frac{d^2u_k(x)}{dx^2} - \frac{i\hbar^2k}{m}\frac{du_k(x)}{dx} + \frac{\hbar^2k^2}{2m}u_k(x) + V(x)u_k(x) = E_ku_k(x)
$$

This can be written as an eigenvalue equation:

$$
\left[-\frac{\hbar^2}{2m}\frac{d^2}{dx^2} - \frac{i\hbar^2k}{m}\frac{d}{dx} + \frac{\hbar^2k^2}{2m} + V(x)\right]u_k(x) = E_ku_k(x)
$$

Alternatively, recognizing the operator form:

$$
\frac{1}{2m}\left(-i\hbar\frac{d}{dx} + \hbar k\right)^2u_k(x) + V(x)u_k(x) = E_ku_k(x)
$$

Thus, $u_k(x)$ is an eigenfunction of the operator:

$$
\hat{H}_k = \frac{1}{2m}\left(-i\hbar\frac{d}{dx} + \hbar k\right)^2 + V(x)
$$

with eigenvalue $E_k$.

The function $u_k(x)$ satisfies periodic boundary conditions:

$$
u_k(x + L) = u_k(x)
$$

Typically, we normalize $u_k(x)$ over one unit cell:

$$
\int_{0}^{L} |u_k(x)|^2 dx = 1
$$

---

We introduce the following dimensionless quantities:

$$
   \begin{cases}
      x' = \frac{x}{L} \\
      k' = kL \quad \text{Since $k$ has dimensions [1/length] and $L$ has dimensions [length], $k'$ is dimensionless} \\
      \epsilon = \frac{\hbar^2}{2mL^2} \quad \text{which has dimensions of energy} \\
      E'_{k'} = \frac{E_{k = k'/ L}}{\epsilon} \\
      V'(x') = \frac{V(x = Lx')}{\epsilon} \\
      u'_{k'}(x') = \sqrt{L} \, u_{k = k'/L}(x = Lx') \quad \text{The factor $\sqrt{L}$ ensures proper normalization in dimensionless coordinates}
   \end{cases}
$$

We transform the differential equation:

$$
-\frac{\hbar^2}{2m}\frac{d^2u_k(x)}{dx^2} - \frac{i\hbar^2k}{m}\frac{du_k(x)}{dx} + \left(\frac{\hbar^2k^2}{2m} + V(x)\right)u_k(x) = E_ku_k(x)
$$

Using the chain rule:

$$
\frac{d}{dx} = \frac{d}{dx'}\frac{dx'}{dx} = \frac{1}{L}\frac{d}{dx'}
$$

$$
\frac{d^2}{dx^2} = \frac{1}{L^2}\frac{d^2}{dx'^2}
$$

Transform each term as shown:

First term:

$$
-\frac{\hbar^2}{2m}\frac{d^2u_k(x)}{dx^2} = -\frac{\hbar^2}{2m} \cdot \frac{1}{L^2}\frac{d^2}{dx'^2}\left(\frac{u'_{k'}(x')}{\sqrt{L}}\right) = -\frac{\hbar^2}{2mL^{5/2}}\frac{d^2u'_{k'}(x')}{dx'^2}
$$

Second term:

$$
-\frac{i\hbar^2k}{m}\frac{du_k(x)}{dx} = -\frac{i\hbar^2}{m}\cdot\frac{k'}{L}\cdot\frac{1}{L}\frac{d}{dx'}\left(\frac{u'_{k'}(x')}{\sqrt{L}}\right) = -\frac{i\hbar^2k'}{mL^{5/2}}\frac{du'_{k'}(x')}{dx'}
$$

Third term:

$$
\frac{\hbar^2k^2}{2m}u_k(x) = \frac{\hbar^2}{2m}\cdot\frac{k'^2}{L^2}\cdot\frac{u'_{k'}(x')}{\sqrt{L}} = \frac{\hbar^2k'^2}{2mL^{5/2}}u'_{k'}(x')
$$

Fourth term:

$$
V(x)u_k(x) = V(Lx')\cdot\frac{u'_{k'}(x')}{\sqrt{L}} = \frac{1}{\sqrt{L}}V(Lx')u'_{k'}(x')
$$

Right-hand side:

$$
E_ku_k(x) = E_k\cdot\frac{u'_{k'}(x')}{\sqrt{L}}
$$

Multiply the entire equation by $\sqrt{L}$ to simplify:

First term: $-\frac{\hbar^2}{2mL^2}\frac{d^2u'_{k'}(x')}{dx'^2}$
Second term: $-\frac{i\hbar^2k'}{mL^2}\frac{du'_{k'}(x')}{dx'}$
Third term: $\frac{\hbar^2k'^2}{2mL^2}u'_{k'}(x')$
Fourth term: $V(Lx')u'_{k'}(x')$
Right-hand side: $E_k u'_{k'}(x')$

Now substitute $V(Lx') = \epsilon V'(x') = \frac{\hbar^2}{2mL^2} V'(x')$ and $E_k = \epsilon E'_{k'} = \frac{\hbar^2}{2mL^2} E'_{k'}$:

First term: $-\frac{\hbar^2}{2mL^2}\frac{d^2u'_{k'}(x')}{dx'^2}$
Second term: $-\frac{i\hbar^2k'}{mL^2}\frac{du'_{k'}(x')}{dx'}$
Third term: $\frac{\hbar^2k'^2}{2mL^2}u'_{k'}(x')$
Fourth term: $\frac{\hbar^2}{2mL^2}V'(x')u'_{k'}(x')$
Right-hand side: $\frac{\hbar^2}{2mL^2}E'_{k'}u'_{k'}(x')$

Now multiply the entire equation by $\frac{2mL^2}{\hbar^2}$:

First term: $-\frac{d^2u'_{k'}(x')}{dx'^2}$
Second term: $-2ik'\frac{du'_{k'}(x')}{dx'}$
Third term: $k'^2 u'_{k'}(x')$
Fourth term: $V'(x')u'_{k'}(x')$
Right-hand side: $E'_{k'}u'_{k'}(x')$

The dimensionless equation is:

$$
\left[-\frac{d^2}{dx'^2} - 2ik'\frac{d}{dx'} + k'^2 + V'(x')\right]u'_{k'}(x') = E'_{k'}u'_{k'}(x')
$$

This can be written more compactly as:

$$
\left[\left(-i\frac{d}{dx'} + k'\right)^2 + V'(x')\right]u'_{k'}(x') = E'_{k'}u'_{k'}(x')
$$

The periodic boundary condition becomes:

$$
u'_{k'}(x' + 1) = u'_{k'}(x')
$$

since when $x$ increases by $L$, $x'$ increases by 1.

The original normalization condition:

$$
\int_{0}^{L} |u_k(x)|^2 dx = 1
$$

Transforms to:

$$
\int_{0}^{L} \left|\frac{u'_{k'}(x')}{\sqrt{L}}\right|^2 dx = \int_{0}^{1} \left|\frac{u'_{k'}(x')}{\sqrt{L}}\right|^2 L dx' = \int_{0}^{1} |u'_{k'}(x')|^2 dx' = 1
$$

Thus, $u'_{k'}(x')$ is normalized over the interval $[0, 1]$ in dimensionless coordinates.

The full Bloch wave function in dimensionless form is:

$$
\psi_k(x) = e^{ikx}u_k(x) = e^{ik'x'}\frac{u'_{k'}(x')}{\sqrt{L}}
$$

If we define $\psi'_{k'}(x') = \sqrt{L}\psi_k(Lx')$, then:

$$
\psi'_{k'}(x') = e^{ik'x'}u'_{k'}(x')
$$

and $\int_0^1 |\psi'_{k'}(x')|^2 dx' = 1$.

Then, the problem to solve is:

$$
\left[\left(-i\frac{d}{dx'} + k'\right)^2 + V'(x')\right]u'_{k'}(x') = E'_{k'}u'_{k'}(x')
$$

For $x \in [0, 1]$ and with periodic boundary conditions $u'_{k'}(x' = 0) = u'_{k'}(x' = 1)$ and normalization $\int_{0}^{1} |u'_{k'}(x')|^2 dx' = 1$.

This dimensionless form eliminates physical constants and scales the problem to unit periodicity, making it ideal for computational implementation

---

Let's write the periodic functions as Fourier series with period unitary:

$$
V'(x') = \sum_{m=-\infty}^{\infty} V_m e^{i2\pi m x'}
$$

where $V_m = \int_0^1 V'(x') e^{-i2\pi m x'} dx'$

$$
u'_{k'}(x') = \sum_{n=-\infty}^{\infty} c_n(k') e^{i2\pi n x'}
$$

where $c_n(k') = \int_0^1 u'_{k'}(x') e^{-i2\pi n x'} dx'$

Let's substituting into the equation, compute $\left(-i\frac{d}{dx'} + k'\right)^2 \phi_n(x')$:

First derivative:

$$
-i\frac{d}{dx'} \phi_n(x') = -i \cdot i2\pi n \phi_n(x') = 2\pi n \phi_n(x')
$$

So:

$$
\left(-i\frac{d}{dx'} + k'\right) \phi_n(x') = (2\pi n + k') \phi_n(x')
$$

Squaring this operator:

$$
\left(-i\frac{d}{dx'} + k'\right)^2 \phi_n(x') = (2\pi n + k')^2 \phi_n(x')
$$

Using the expansion:

$$
\left(-i\frac{d}{dx'} + k'\right)^2 u'_{k'}(x') = \sum_{n=-\infty}^{\infty} c_n(k') (2\pi n + k')^2 e^{i2\pi n x'}
$$

The potential term is:

$$
V'(x')u'_{k'}(x') = \left(\sum_{m=-\infty}^{\infty} V_m e^{i2\pi m x'}\right) \left(\sum_{n=-\infty}^{\infty} c_n(k') e^{i2\pi n x'}\right)
$$

Combine the sums:

$$
V'(x')u'_{k'}(x') = \sum_{m,n} V_m c_n(k') e^{i2\pi (m+n) x'}
$$

Let $p = m + n$, so $n = p - m$:

$$
V'(x')u'_{k'}(x') = \sum_{p=-\infty}^{\infty} \left(\sum_{m=-\infty}^{\infty} V_m c_{p-m}(k')\right) e^{i2\pi p x'}
$$

Putting it all together:

$$
\sum_{n=-\infty}^{\infty} (2\pi n + k')^2 c_n(k') e^{i2\pi n x'} + \sum_{p=-\infty}^{\infty} \left(\sum_{m=-\infty}^{\infty} V_m c_{p-m}(k')\right) e^{i2\pi p x'} = E'_{k'} \sum_{n=-\infty}^{\infty} c_n(k') e^{i2\pi n x'}
$$

Since $e^{i2\pi n x'}$ are orthogonal, equate coefficients for each $n$:

$$
(2\pi n + k')^2 c_n(k') + \sum_{m=-\infty}^{\infty} V_m c_{n-m}(k') = E'_{k'} c_n(k')
$$

This is an infinite set of equations. Let's change indices: let $m' = n - m$, then $m = n - m'$:

$$
(2\pi n + k')^2 c_n(k') + \sum_{m'=-\infty}^{\infty} V_{n-m'} c_{m'}(k') = E'_{k'} c_n(k')
$$

We truncate to $N$ modes: $n = -N, -N+1, \ldots, 0, \ldots, N-1, N$

Define:
- Vector $\mathbf{c}(k') = [c_{-N}(k'), c_{-N+1}(k'), \ldots, c_0(k'), \ldots, c_N(k')]^T$
- Hamiltonian matrix $H(k')$ with elements:

$$
H_{n,n}^{\text{kinetic}} = (2\pi n + k')^2 \quad \text{it's a diagonal matrix}
$$

$$
H_{n,m}^{\text{potential}} = V_{n-m} \quad \text{it's a Toeplitz matrix}
$$

where $V_{n-m}$ is the Fourier coefficient of the potential

The equation becomes:

$$
\boxed{H(k') \mathbf{c}(k') = E'_{k'} \mathbf{c}(k')}
$$

Where $H(k') = H^{\text{kinetic}}(k') + H^{\text{potential}}$

Some details:

- The matrix is $(2N+1) \times (2N+1)$.
- Since $V'(x')$ is real, $V_{-m} = V_m^*$. This ensures $H(k')$ is Hermitian
- From Parseval's theorem:

$$
\int_0^1 |u'_{k'}(x')|^2 dx' = \sum_{n=-N}^{N} |c_n(k')|^2 = 1
$$

So eigenvectors should be normalized

- $k'$ is in $[-\pi, \pi]$, but we can use symmetry: $E'_n(k') = E'_n(-k')$

Let's solve the eigenvalue problem for a gaussian periodic potential

```python
# Importing libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import rfft
from scipy.linalg import eigvalsh, toeplitz
```

```python
# Define the potential function
V0 = 30.0
center = 0.5
sigma = 0.1

V = lambda x: - V0 * np.exp(- (x - center) ** 2 / (2 * sigma ** 2))
```

```python
# Sample points for FFT
N_samples = 1_000
x_sampled = np.linspace(0, 1, N_samples, endpoint = False)
V_sampled = V(x_sampled)
```

```python
N = 20  # Truncation parameter: modes from - N to N

# Compute Fourier coefficients using rfft() from Scipy
V_coeffs_rfft = rfft(V_sampled) / N_samples
V_pos = V_coeffs_rfft.real[0 : 2 * N + 1 : 1] # .real to remove imaginary noise
```

```python
# Raw coefficients
V_coeffs_rfft
```

```python
# values for the potential_matrix
V_pos
```

```python
# k' values from - pi to pi
k_points = 200  # Number of k' values
k_vals = np.linspace(- np.pi, np.pi, k_points)

# Pre-compute the potential matrix
n_vals = np.arange(- N, N + 1, 1)
potential_matrix = toeplitz(V_pos)  # Toeplitz from V_0 to V_{2N} and V_0 to V_{-2N}
potential_matrix
```

```python
# Initialize array for eigenvalues
eigenvalues = np.zeros((k_points, 2 * N + 1))

# Loop over k' values
for i, k_prime in enumerate(k_vals):

    kinetic_diag = (2 * np.pi * n_vals + k_prime) ** 2 # Kinetic term (diagonal)
    H = np.diag(kinetic_diag) + potential_matrix # Hamiltonian matrix
    
    # Diagonalize (eigenvalues only)
    eigenvalues[i, :] = np.sort(eigvalsh(H))

eigenvalues
```

```python
# Plotting
_, (ax1, ax2) = plt.subplots(1, 2, figsize = (12, 5))

n_bands_plot = 4

# Plot band structure (first bands)
for band in range(n_bands_plot):
    ax1.plot(k_vals, eigenvalues[:, band], linewidth = 2, label = f'Band {band}')
ax1.set_xlabel(r"$k \, \left[ \frac{1}{L} \right]$", fontsize = 12)
ax1.set_ylabel(r"$E \, \left[ \frac{\hbar ^ 2}{2 m L ^ 2} \right]$", fontsize = 12)
ax1.set_title(f"Band Structure (First {n_bands_plot} Bands)", fontsize = 14)
ax1.grid(True, alpha = 0.5)
ax1.legend(loc = 'upper right')
ax1.set_xlim(- np.pi, np.pi)

# Plot potential
ax2.plot(x_sampled, V_sampled, 'k-', linewidth = 2)
ax2.set_xlabel(r"$x \, \left[ \frac{1}{L} \right]$", fontsize = 12)
ax2.set_ylabel(r"$V \left( \frac{x}{L} \right) \, \left[ \frac{\hbar ^ 2}{2 m L ^ 2} \right]$", fontsize = 12)
ax2.set_title("Gaussian Hole Potential", fontsize = 14)
ax2.grid(True, alpha = 0.5)
ax2.set_xlim(0, 1)

plt.tight_layout()
plt.show()

print(f"\nFirst band (n=0) energy range:")
print(f"  Minimum: {np.min(eigenvalues[:, 0]):.4f} at k' = {k_vals[np.argmin(eigenvalues[:, 0])]:.4f}")
print(f"  Maximum: {np.max(eigenvalues[:, 0]):.4f} at k' = {k_vals[np.argmax(eigenvalues[:, 0])]:.4f}")
print(f"  Band width: {np.max(eigenvalues[:, 0]) - np.min(eigenvalues[:, 0]):.4f}")

print(f"\nBand gaps:")
for band in range(n_bands_plot):
    gap = np.min(eigenvalues[:, band + 1]) - np.max(eigenvalues[:, band])
    print(f"  Between band {band} and {band + 1}: {gap:.4f}")
```
