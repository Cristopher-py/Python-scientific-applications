```python
import sympy as smp
```

```python
a1, a2, a3 = smp.symbols("a1:4", complex = True)
b1, b2, b3, b4, b5 = smp.symbols("b1:6", complex = True)

a = smp.Matrix([a1, a2, a3])
b = smp.Matrix([b1, b2, b3, b4, b5])
```

The outer product of two vectors $\mathbf{a} \in \mathbb{C}^m$ and $\mathbf{b} \in \mathbb{C}^n$ is defined as:

$$
    \mathbf{a} \mathbf{b}^\dagger
$$

or equivalently,

$$
    \mathbf{a} \otimes \mathbf{b}^\dagger
$$

where $\mathbf{b}^\dagger$ denotes the conjugate transpose (Hermitian adjoint) of $\mathbf{b}$.

Let

$$
    \mathbf{a} = \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_m \end{bmatrix}, \quad
    \mathbf{b} = \begin{bmatrix} b_1 \\ b_2 \\ \vdots \\ b_n \end{bmatrix}
$$

then

$$
    \mathbf{b}^\dagger = \begin{bmatrix} \bar{b}_1 & \bar{b}_2 & \dots & \bar{b}_n \end{bmatrix}
$$

and the outer product is:

$$
    \mathbf{a} \mathbf{b}^\dagger =
        \begin{bmatrix}
            a_1 \bar{b}_1 & a_1 \bar{b}_2 & \dots & a_1 \bar{b}_n \\
            a_2 \bar{b}_1 & a_2 \bar{b}_2 & \dots & a_2 \bar{b}_n \\
            \vdots & \vdots & \ddots & \vdots \\
            a_m \bar{b}_1 & a_m \bar{b}_2 & \dots & a_m \bar{b}_n
        \end{bmatrix}
$$

which is an $m \times n$ complex matrix with the $(i,j)$ entry being $a_i \bar{b}_j$.

```python
# ab.H
outer_ab = a @ b.H
outer_ab
```

```python
# ba.H
outer_ba = b @ a.H
outer_ba
```

```python
# ab.H acts on vectors in the same b's space
outer_ab @ b
```

```python
# ba.H acts on vectors in the same a's space
outer_ba @ a
```

```python
# outer product is not commutative
outer_ab.equals(outer_ba)
```

```python
# This holds
outer_ab.equals(outer_ba.H)
```

```python
# This does not hold
(outer_ab @ b).equals(outer_ba @ a)
```

Kronecker product of two vectors $\mathbf{a} \in \mathbb{C}^m$ and $\mathbf{b} \in \mathbb{C}^n$ is defined as:

$$
    \mathbf{a} \otimes \mathbf{b}
$$

Let

$$
    \mathbf{a} = \begin{bmatrix} a_1 \\ a_2 \\ \vdots \\ a_m \end{bmatrix}, \quad
    \mathbf{b} = \begin{bmatrix} b_1 \\ b_2 \\ \vdots \\ b_n \end{bmatrix}
$$

then

$$
    \mathbf{a} \otimes \mathbf{b} =
        \begin{bmatrix}
            a_1 \mathbf{b} \\
            a_2 \mathbf{b} \\
            \vdots \\
            a_m \mathbf{b}
    \end{bmatrix}
=
    \begin{bmatrix}
        a_1 b_1 \\
        a_1 b_2 \\
        \vdots \\
        a_1 b_n \\
        a_2 b_1 \\
        \vdots \\
        a_m b_n
    \end{bmatrix}

$$

which is a column vector of size $mn \times 1$.

```python
# a (x) b
axb = smp.kronecker_product(a, b)
axb
```

```python
# b (x) a
bxa = smp.kronecker_product(b, a)
bxa
```

```python
# Kronecker product is not commmutative
axb.equals(bxa)
```

```python
# Dot product
axb.dot(bxa, hermitian = True)
```

```python
# Outer product axb_bxa.H
outer_k = axb @ bxa.H
outer_k
```

```python
# outer_k is not hermitian
outer_k.H.equals(outer_k)
```

```python
# outer_k has a zero eigenvalue
outer_k.det()
```

```python
# The conjugate of the outer_k's trace is equal the hermitian dot product between a (x) b and b (x) a
smp.conjugate(outer_k.trace()).equals(axb.dot(bxa, hermitian = True))
```
