### n-th root of a number

Halley’s method computes a root of a function $f(x)$ by applying:


\begin{cases}
    x_{i+1} = x_i - \dfrac{2 f(x_i) f'(x_i)}{2[f'(x_i)]^2 - f(x_i) f''(x_i)} \\
    x_0 \in [a,b]
\end{cases}

If we apply this to:

$$
    f(x) = x^n - k
$$

We obtain an iterative formula to get the $n$-th root of $k$:

$$
    x_{i+1} = x_i \, \frac{(n-1)x_i^n + (n+1)k} {(n+1)x_i^n + (n-1)k}
$$

For the square root, i.e. $n=2$, this becomes:

$$
    x_{i+1} = x_i \, \frac{x_i^2 + 3k}{3x_i^2 + k}
$$

This is a cubically convergent variant of Heron’s formula. Halley’s method requires fewer iterations to reach machine precision. Let's implement it using a simple stopping criterion based the distance between 2 consecutive machine numbers

```python
from math import ulp  # Distance between 2 machine numbers
import numpy as np
```

There're 2 scenarios: either $x_{i + 1}$ can be exactly equal to $x_{i}$, or $x_{i + 1}$ oscillates between 2 machine numbers

```python
def root_arr(k, n, x0, max_iter = 50): # Just for analysis

    x_old = x0
    x_array = np.zeros(max_iter + 1)
    x_array[0] = x0
    
    for j in range(1, max_iter + 1, 1):

        x_new = x_old * (x_old ** n * (n - 1.0) + k * (n + 1.0)) / (x_old ** n * (n + 1.0) + k * (n - 1.0)) # Basic step
        x_array[j] = x_new # Append result
        x_old = x_new # Update x_old

    return x_array # Returns the array
```

```python
# Cubic root of 13
k = 13.0
n = 3.0
x0 = 1.0

x = root_arr(k, n, x0) # Convergence is fast
x
```

Let's handle these cases

```python
def n_root(k, n, x0, max_iter = 50, verbose = True): # Implements Halley's method with estimated error

    step = lambda x: x * (x ** n * (n - 1.0) + k * (n + 1.0)) / (x ** n * (n + 1.0) + k * (n - 1.0)) # Basic step

    x_old = x0
    
    for j in range(1, max_iter + 1, 1):

        x_new = step(x_old) # One step
        diff = abs(x_new - x_old); error = ulp(x_new) # The error is 1 ULP

        if diff <= 2.0 * error: # Let's find a fixed value (if any)

            if diff == 0.0: # Fixed value suddenly reached
                if verbose: print("Fixed value detected, 1st shot")
                return (x_new, error, j)
            
            else: # Let's see forward, just give it a possibility
                x_new2 = step(x_new); x_new3 = step(x_new2) # 2 more steps
                diff = abs(x_new3 - x_new2); error = ulp(x_new3)

                if diff == 0.0: # Fixed value found
                    if verbose: print("Fixed value detected, 2nd shot")
                    return (x_new3, error, j + 2)
                else: # Oscillatory behaviour, no fixed value detected
                    if verbose: print("Oscillatory behaviour")
                    m = (x_new3 + x_new2) / 2.0 # Middle point
                    return (m, ulp(m), j + 2)

        x_old = x_new # Update x_old

    if verbose: print('Maximum number of iterations reached. The results are not reliable')
    return (x_new, error, j)
```

```python
# Cubic root of 13
k = 13.0
n = 3.0
x0 = 1.0

n_root(k, n, x0), k ** (1 / n) # Oscillating
```

```python
# Cubic root of 188
k = 188.0
n = 3.0
x0 = 1.0

n_root(k, n, x0), k ** (1 / n) # Oscillating
```

```python
# Cubic root of 13
k = 60.0
n = 5.0
x0 = 1.0

n_root(k, n, x0), k ** (1 / n) # Fixed value
```

Let's compute a table for $\sqrt[3]{k}$

```python
for k in range(2, 216 + 1, 1): # Cubic root of all natural numbers between 2 and 215

    x0 = k / 20.0 # Initial value
    x, err, j = n_root(k, 3, x0, verbose = False)
    real = k ** (1 / 3)
    err_real = ulp(real) # Estimated roundoff error for k ** (1 / 3)

    print(f"\u221b{k} = {x:.16f} +- {err:.1g} ({j}) \t  {real:.16f} +- {err_real:.1g}")
```

Let's compute a table for $\sqrt[4]{k}$

```python
for k in range(2, 256 + 1, 1): # Fourth root of all natural numbers between 2 and 256

    x0 = k / 20.0 # Initial value
    x, err, j = n_root(k, 4, x0, verbose = False)
    real = k ** (1 / 4)
    err_real = ulp(real) # Estimated roundoff error for k ** (1 / 4)

    print(f"\u221c{k} = {x:.16f} +- {err:.1g} ({j}) \t  {real:.16f} +- {err_real:.1g}")
```
