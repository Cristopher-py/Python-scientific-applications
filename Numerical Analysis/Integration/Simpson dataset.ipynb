### Error estimation of an integral, Simpson rule

Given a set of $N$ points $y_i$ (analytic function is unknown), the integral $I_N$ can be computed using Simpson rule. An estimation of the error might be:

$$
    E = \frac{\left| 1024 \, I_{2N} - 1025 \, I_N + I_{\frac{N}{2}} \right|}{945} \qquad \text{Richardson estimation}
$$

Error model:

$$
    I_{true} = I(N) + \frac{C_1}{N ^ 4} + \frac{C_2}{N ^ 6}
$$

```python
from itertools import pairwise
from scipy.integrate import simpson
```

```python
y = [12.5, 13.1, 11.7, 9.3, 8.3, 6.3, 5.3, 4.6, 5.1, 6.4, 10.3, 13.9] # Original points, N = 12
y_2 = y[::2] # Subset, N = 6
y2 = [v for pair in pairwise(y) for v in (pair[0], sum(pair) / 2)] + [y[-1]] # Extended list with midpoints, N = 23
```

```python
I = simpson( # I(N)
    y = y,
    dx = 1.0
)

I_2 = simpson( # I(N / 2)
    y = y_2,
    dx = 2.0
)

I2 = simpson( # I(2 * N)
    y = y2,
    dx = 0.5
)

I, I_2, I2 # Results
```

```python
# I and estimated absolute error using Richardson method
I, abs(1024.0 * I2 - 1025.0 * I + I_2) / 945.0
```

Result:

$$
    I = 92.9 \pm 0.7
$$

```python
# Relative error [%]
(0.7 / 92.9) * 100.0
```
