### Romberg method

See in *PDF* folder for precise symbolic expressions

```python
import numpy as np
from scipy.integrate import quad
```

```python
def integral(f, a, b, h): # Order 2 integration method

    ba = b - a
    N = round(ba / h)
    h = ba / N # Recalculate h to match the exact grid
    x = np.linspace(a, b, N, endpoint = False)
    midpoints = x + h / 2.0

    I = h * np.sum(f(midpoints))

    return I
```

```python
f = lambda x: np.sqrt(x * np.tan(x)) # f(x) = sqrt(x * tan(x))
a, b = 0.0, np.pi / 3.0 # a = 0, b = pi / 3

h = 1e-02 # h = 0.01, not so small
N = 5
Ih = []
for _ in range(N + 1): # Computing I(h), I(h/2), I(h/4), ...
    x = np.arange(a, b + h, h)
    Ih.append(
        integral(
            f = f,
            a = a,
            b = b,
            h = h
        )
    )

    h /= 2.0

Ih
```

```python
I_romb1 = (- Ih[0] + 4.0 * Ih[1]) / 3.0 # One extrapolation
I_romb1, abs(I_romb1 - Ih[0])
```

```python
I_romb2 = (Ih[0] - 20.0 * Ih[1] + 64.0 * Ih[2]) / 45.0 # Two extrapolations
I_romb2, abs(I_romb2 - I_romb1), abs(I_romb2 - Ih[0])
```

```python
I_romb3 = (- Ih[0] + 84.0 * Ih[1] - 1344.0 * Ih[2] + 4096.0 * Ih[3]) / 2835.0 # Three extrapolations
I_romb3, abs(I_romb3 - I_romb2), abs(I_romb3 - Ih[0])
```

```python
I_romb4 = (Ih[0] - 340.0 * Ih[1] + 22848.0 * Ih[2] - 348160.0 * Ih[3] + 1048576.0 * Ih[4]) / 722925.0 # Four extrapolations
I_romb4, abs(I_romb4 - I_romb3), abs(I_romb4 - Ih[0])
```

```python
real, err = quad(f, a, b) # Precise value
real, err
```

```python
# Romb4 is much better
abs(I_romb4 - real), abs(Ih[0] - real)
```
