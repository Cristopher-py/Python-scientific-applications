### Romberg method

See in *PDF* folder for precise symbolic expressions

```python
import numpy as np
from scipy.integrate import quad
```

```python
def integral(f, a, b, h): # Order 2 integration method

    ba = b - a
    N = round(ba / h)
    h = ba / N # Recalculate h to match the exact grid
    x = np.linspace(a, b, N, endpoint = False)
    midpoints = x + h / 2.0

    I = h * np.sum(f(midpoints))

    return I
```

```python
f = lambda x: np.sqrt(x * np.tan(x)) # f(x) = sqrt(x * tan(x))
a, b = 0.0, np.pi / 3.0 # a = 0, b = pi / 3

h = 1e-02 # h = 0.01, not so small
N = 6
Ih = []
for _ in range(N + 1): # Computing I(h), I(h/2), I(h/4), ...
    x = np.arange(a, b + h, h)
    Ih.append(
        integral(
            f = f,
            a = a,
            b = b,
            h = h
        )
    )

    h /= 2.0

Ih
```

Do just 5 extrapolations. Take $I_{romb}^{4}$ as the best result. The final absolute error can be estimated as $ \left| I_{romb}^{5} - I_{romb}^{4} \right|$

```python
I_romb1 = (- Ih[0] + 4.0 * Ih[1]) / 3.0 # One extrapolation
I_romb1
```

```python
I_romb2 = (Ih[0] - 20.0 * Ih[1] + 64.0 * Ih[2]) / 45.0 # Two extrapolations
I_romb2, abs(I_romb2 - I_romb1) # Absolute error of I_romb1
```

```python
I_romb3 = (- Ih[0] + 84.0 * Ih[1] - 1344.0 * Ih[2] + 4096.0 * Ih[3]) / 2835.0 # Three extrapolations
I_romb3, abs(I_romb3 - I_romb2) # Absolute error of I_romb2
```

```python
I_romb4 = (Ih[0] - 340.0 * Ih[1] + 22848.0 * Ih[2] - 348160.0 * Ih[3] + 1048576.0 * Ih[4]) / 722925.0 # Four extrapolations
I_romb4, abs(I_romb4 - I_romb3) # I_romb4 is our best result. # abs(I_romb4 - I_romb3) is the absolute error of I_romb3
```

```python
I_romb5 = (- Ih[0] + 1364.0 * Ih[1] - 371008.0 * Ih[2] + 23744512.0 * Ih[3] - 357564416.0 * Ih[4] + 1073741824.0 * Ih[5]) / 739552275.0 # Five extrapolations
I_romb5, abs(I_romb5 - I_romb4) # Absolute error of I_romb4
```

```python
real, err = quad(f, a, b) # Precise value
real, err
```

```python
# Romb4 is much better than I_h
abs(I_romb4 - real), abs(Ih[0] - real)
```

Best Romberg result computed:

$$
    \int_0^{\frac{\pi}{3}} \, \sqrt{x \, \tan{x}} \, dx = 0.614876900602 \pm 0.000000000007
$$

Let's reduce $h$

```python
h = 1e-03 # h = 0.001
Ih = []
for _ in range(N + 1): # Computing I(h), I(h/2), I(h/4), ...
    x = np.arange(a, b + h, h)
    Ih.append(
        integral(
            f = f,
            a = a,
            b = b,
            h = h
        )
    )

    h /= 2.0

Ih
```

```python
I_romb4 = (Ih[0] - 340.0 * Ih[1] + 22848.0 * Ih[2] - 348160.0 * Ih[3] + 1048576.0 * Ih[4]) / 722925.0 # Four extrapolations
I_romb5 = (- Ih[0] + 1364.0 * Ih[1] - 371008.0 * Ih[2] + 23744512.0 * Ih[3] - 357564416.0 * Ih[4] + 1073741824.0 * Ih[5]) / 739552275.0 # Five extrapolations

I_romb4, abs(I_romb5 - I_romb4) # Result, error
```

Result:

$$
    \int_0^{\frac{\pi}{3}} \, \sqrt{x \, \tan{x}} \, dx = 0.61487690059894 \pm 0.00000000000003 
$$

```python
# Real error committed
abs(I_romb4 - real)
```

Let's reduce $h$ again. See what happens

```python
h = 25e-07 # h = 0.0000025
Ih = []
for _ in range(N + 1): # Computing I(h), I(h/2), I(h/4), ...
    x = np.arange(a, b + h, h)
    Ih.append(
        integral(
            f = f,
            a = a,
            b = b,
            h = h
        )
    )

    h /= 2.0

I_romb4 = (Ih[0] - 340.0 * Ih[1] + 22848.0 * Ih[2] - 348160.0 * Ih[3] + 1048576.0 * Ih[4]) / 722925.0 # Four extrapolations
I_romb5 = (- Ih[0] + 1364.0 * Ih[1] - 371008.0 * Ih[2] + 23744512.0 * Ih[3] - 357564416.0 * Ih[4] + 1073741824.0 * Ih[5]) / 739552275.0 # Five extrapolations

I_romb4, abs(I_romb5 - I_romb4) # Result, error
```

Result:

$$
    \int_0^{\frac{\pi}{3}} \, \sqrt{x \, \tan{x}} \, dx = 0.6148769005989663 \pm 0.0000000000000001
$$
