### Central finite difference absolute error

$f'(x)$ can be computed using the formula (numerical approximation):

$$
    f'(x) = \frac{f(x + h) - f(x - h)}{2h}
$$

The error bound carried is given by the sum of all contributions:

$$
    E \lesssim \frac{1}{h} \varepsilon_{mach} \left| f(x) \right| + \varepsilon_{mach} \left| x \right| \left| f''(x) \right| + \frac{h^2}{6} \left| f'''(x) \right|
$$

- $\frac{1}{h} \varepsilon_{mach} \left| f(x) \right|$ is the roundoff error from function evaluations;
- $\varepsilon_{mach} \left| x \right| \left| f''(x) \right|$ is the propagated error from the inaccurate representation of $x$;
- $\frac{h^2}{6} \left| f'''(x) \right|$ is the truncation error from the finite difference approximation;

The best value of $h$ that minimizes the error is:

$$
    h_{opt} = \sqrt[3]{\frac{3 \, \varepsilon_{mach} \, \left| f(x) \right|}{\left| f'''(x) \right|}}
$$

Let's see how much precise can we be with it

```python
import numpy as np

eps = 2.2e-16 # Machine precision for doubles
```

```python
f = lambda x: x ** (1 / 3) # Test: x^(1 / 3)
f2 = lambda x: - 2.0 / 9.0 * (x ** (- 5 / 3)) # Second derivative
f3 = lambda x: 10.0 / 27.0 * (x ** (- 8 / 3)) # Third derivative

def derivative(f, x, h): # Applies the central finite difference formula
    return (f(x + h) - f(x - h)) / (2.0 * h)

def error(f, f2, f3, x, h): # Compute the error bound
    global eps
    return 1.0 / h * eps * np.abs(f(x)) + eps * np.abs(x) * np.abs(f2(x)) + h ** 2 / 6.0 * np.abs(f3(x))
```

```python
x = 1.5 # Point target
h = 1e-08 # Initial step 10^{-8}

derivative(f, x, h), error(f, f2, f3, x, h)
```

Let's compute the best value for $h$

```python
h_opt = (3 * eps * np.abs(f(x)) / np.abs(f3(x))) ** (1 / 3)
h_opt
```

As we can see, $h$ is not as small as we thought intuitively. Surprisingly, if $h < 10^{-5}$ the error increases!

```python
derivative(f, x, h_opt), error(f, f2, f3, x, h_opt) # Best result
```

Our result is:

$$
    f' \left( \frac{3}{2} \right) = 0.25438094280 \pm 0.00000000002
$$
