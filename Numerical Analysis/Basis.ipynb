### Basis Representation Theorem

Every non-zero real number $x$ can be written uniquely as:

\begin{equation}
    \boxed{x = \text{sgn}(x) \, \beta^{p} \, \sum_{k = 1}^{+ \infty} \, d_{k} \, \beta^{-k}}
\end{equation}

Where:

- $\beta$ is the base, a natural number greater than or equal to 2
- $p$ is the exponent, an integer number (which can be zero as well)
- Each $d_k$ is a natural number less than the base $\beta$, and $d_1$ is not zero
- $\sum_{k = 1}^{+ \infty} \, d_{k} \, \beta^{-k}$ is the mantissa, a number confined within 1 / $\beta$ (strictly) and 1

The sequence $d_1, d_2, d_3, \dots$ \, does not eventually become constantly equal to $\beta$ - 1

Let's see some fundamental basis transformations

### Base convertion: integer case

Let x be an integer number with $N$ digits $\pm \, d_1 d_2 d_3 \dots d_N$. We can write it using the Theorem:

$$
    x = \text{sgn(x)} \, \sum_{k = 1}^{N} \, d_{k} \, \beta^{N - k}
$$

We can isolate d_{N}, the last digit, as shown:

$$
    x = \text{sgn(x)} \, \left( \sum_{k = 1}^{N - 1} \, d_{k} \, \beta^{N - k} + d_{N} \right)
$$

Factor out $\beta$ to get:

$$
    x = \text{sgn(x)} \, \left[ \left( \sum_{k = 1}^{N - 1} \, d_{k} \, \beta^{N - k - 1} \right) \beta + d_{N} \right]
$$

In this form, we can easily see that $d_N$ is the remainder of the division of x by $\beta$. Therefore, $d_{N - 1}$ is the remainder of the division of x by $\beta^{2}$ and so on. We can thus define a basic algorithm to write an integer number in a different base $\beta$

```python
def integer_converter_10_base(number, base): # It converts an integer number from base 10 to beta
    reminders = []
    while number != 0: # Algorithm ends when we reach zero
        r = number % base
        reminders.insert(0, r) # r is inserted at the first position in the list
        number = number // base # Update number: x / beta (integer division)

    return reminders
```

```python
x = 1234 # Number to convert

for beta in range(2, 16): # Choose different basis
    print(f"{x} in base {beta}: {integer_converter_10_base(x, beta)}")
```

We can indeed define an inverse algorithm. Just remember that:

$$
    d_{1} \, d_{2} \, d_{3} \, \dots \, d_{N} = \sum_{k = 1}^{N} \, d_k \, \beta^{N - k}
$$

```python
def integer_converter_base_10(number_list, base): # It converts an integer number from base beta to base 10
    conv = [base ** j for j in range(len(number_list) - 1, - 1, - 1)]
    return sum(a * b for a, b in zip(number_list, conv))
```

```python
integer_converter_base_10( # Demonstration: we should retrieve 1234
    number_list = [5, 4, 1, 4], # This is 1234 written in base 6
    base = 6
)
```

### Base convertion: decimal case

Let $x$ be a real number confined in (0, 1). By the Theorem, we can write:

$$
    x = \sum_{k = 1}^{+ \infty} \, d_{k} \, \beta^{- k}
$$

To isolate the first digit, multiply $x$ by the base $\beta$. We obtain:

$$
    \beta \, x = \sum_{k = 1}^{+ \infty} \, d_{k} \, \beta^{- k + 1} = d_{1} + \sum_{k = 2}^{+ \infty} \, d_{k} \, \beta^{- k + 1}
$$

We can easily see that $d_{1}$ is the integer part of $\beta \, x$. Therefore, $d_{2}$ is the integer part of $\beta^{2} \, x$ and so on. Let's write an algorithm to convert  a real number (confined between 0 and 1) from the base 10 to the base $\beta$

```python
def decimal_converter_10_base(number, base, precision = 32): # It converts a real number between 0 and 1 from base 10 to base beta

    digits = []
    counter = 0 # While-cycle guard

    while (number != 0) and (counter < precision): # Precision: we cannot compute infinite digits
        number *=  base
        integer = int(number)
        digits.append(str(integer))
        number -= integer

        counter += 1

    return "0." + "".join(digits)
```

```python
from numpy import pi, e

for number in [number / 10.0 for number in range(1, 10, 1)]: # Demonstration with rational numbers (0.1, 0.2, ..., 0.9)
    print(f"{number} in base 2: {decimal_converter_10_base(number, 2)}")

print(f"\npi in base 2: {decimal_converter_10_base(pi - 3, 2)}") # pi - 3 is irrational and lies between 0 and 1
print(f"\ne in base 2: {decimal_converter_10_base(e - 2, 2)}") # e - 2 is irrational lies between 0 and 1
```

Here's something fascinating: every rational number between 0 and 1 has a binary expansion that eventually becomes periodic. Irrational numbers, however, have non‑repeating infinite expansions. Since computers cannot work with infinitely many digits (only mathematics can) they must round or truncate these expansions to a fixed number of bits (typically 32 or 64, corresponding to about 7 or 16 decimal digits of precision). This finite representation within computers is one of the primary sources of numerical errors.

```python
number = 0.1234 # Fix a number

for beta in range(2, 11): # Represents the number in different basis
    print(f"{number} in base {beta}: {decimal_converter_10_base(number, beta)}")
```

The algorithm doesn't yield 0.1234 perfectly when converting 0.1234 from base 10 to itself (an identity transformation), because numbers are represented with finite precision
