## Second derivative

By the forward difference method, the second derivative is computed as:

$$
    f''(x) \approx \frac{f(x + 2 \, h) - 2 \, f(x + h) + f(x)}{h^2}
$$

The maximum error committed is:

$$
    E = | f'''(x) | \left( | x | \frac{\varepsilon}{2} + h \right) + \frac{2 \, | f(x) | \, \varepsilon}{h^2}
$$

Where $\varepsilon$ is the machine precision. For doubles, $\varepsilon = 2^{-52} \approx 2.2 \times 10^{-16}$.

The optimal value of $h$ that minimizes the error is:

$$
    h_{opt} = \sqrt[3]{\, 4 \, \varepsilon \, \left| \frac{f(x)}{f'''(x)} \right|}
$$

```python
import numpy as np
```

```python
def second_derivative_forward(x0, h, bo = False):

    f = lambda x: np.exp(- x ** 2) # Target function: e^(-x^2)
    f3 = lambda x: 4.0 * (3.0 - 2.0 * x ** 2) * x * f(x) # Third derivative

    eps = 2.2e-16 # Machine precision

    if bo:
        h = np.cbrt(4.0 * eps * np.abs(f(x0) / f3(x0))) # Best value for h

    second_derivative = (f(x0 + 2.0 * h) - 2.0 * f(x0 + h) + f(x0)) / (h ** 2)
    E = np.abs(f3(x0)) * (np.abs(x0) * eps / 2.0 + h) + (2.0 * np.abs(f(x0))) * eps / (h ** 2)

    return h, second_derivative, E # It returns h, the second derivative and the absolute error
```

```python
np.array([second_derivative_forward(x0 = 1.0, h = h) for h in (10.0 ** (- np.arange(1, 10 + 1, 1)))]) # Values for h = 0.1, 0.01, 0.001, ...
```

```python
second_derivative_forward(x0 = 1.0, h = 0.0, bo = True) # Best result (forward method)
```

Our best result is:

$$
    f''(1) = 0.73577 \pm 0.00001
$$

By central difference, the second derivative can be computed as:

$$
    f''(x) \approx \frac{f(x + h) - 2 \, f(x) + f(x - h)}{h^2}
$$

The upper bound error is:

$$
    E = \frac{2 \, | f(x) | \, \varepsilon}{h^2} + | x \, f'''(x) | \, \frac{\varepsilon}{2} + \frac{h^2}{6} \, \left| f^{iv}(x) \right|
$$

The best value of $h$ is:

$$
    h_{opt} = \sqrt[4]{12 \, \varepsilon \, \left| \frac{f(x)}{f^{iv}(x)} \right|}
$$

```python
def second_derivative_central(x0, h, bo = False):

    f = lambda x: np.exp(- x ** 2) # Target function: e^(-x^2)
    f3 = lambda x: 4.0 * (3.0 - 2.0 * x ** 2) * x * f(x) # Third derivative
    f4 = lambda x: 4 * (4.0 * x ** 4 - 12.0 * x ** 2 + 3.0) * f(x) # Fourth derivative

    eps = 2.2e-16 # Machine precision

    if bo:
        h = np.sqrt(np.sqrt(12.0 * eps * np.abs(f(x0) / f4(x0)))) # Best value for h

    second_derivative = (f(x0 + h) - 2 * f(x0) + f(x0 - h)) / (h ** 2)
    E = h ** 2 / 6.0 * np.abs(f4(x0)) + eps / 2.0 * np.abs(x0 * f3(x0)) + 2 * eps * np.abs(f(x0)) / h ** 2

    return h, second_derivative, E # It returns h, the second derivative and the absolute error
```

```python
np.array([second_derivative_central(x0 = 1.0, h = h) for h in (10.0 ** (- np.arange(1, 10 + 1, 1)))]) # Values for h = 0.1, 0.01, 0.001, ...
```

```python
second_derivative_central(x0 = 1.0, h = 0.0, bo = True) # Best result (central method)
```
