### $e$ Taylor expansion error

Let $x$ be a positive number. Computing:

$$
    e^{x} \approx \sum_{k = 0}^{N} \, \frac{x^{k}}{k!}
$$

The total absolute upper bound error is:

$$
    E_{pos} \lesssim e^{x} \left[ \frac{x^{N + 1}}{\left( N + 1 \right)!} + u \left( 2 N + 1 + x \right) \right]
$$

Where $u$ is the machine precision.

Let $x$ be a negative number. Then we can compute:

$$
    e^{x} = \frac{1}{e^{|x|}} \approx \frac{1}{\sum_{k = 0}^{N} \, \frac{|x|^{k}}{k!}}
$$

And the error is:

$$
    E_{neg} = \frac{E_{pos}(|x|)}{ \left( \sum_{k = 0}^{N} \, \frac{|x|^{k}}{k!}  \right)^{2}} + \frac{u}{\sum_{k = 0}^{N} \, \frac{|x|^{k}}{k!}}
$$

```python
import numpy as np
from scipy.special import factorial
```

```python
eps = 2 ** (- 53) # Machine epsilon

def taylor(x, N): # It sums N terms to compute e^x

    global eps

    x_ = np.abs(x)
    k = np.arange(0, N + 1, 1)
    terms = x_ ** k / factorial(k)

    res = np.sum(terms)
    err = res * (x_ ** (N + 1) / factorial(N + 1) + eps * (x_ + 2.0 * N + 1.0)) # Error formula

    if x >= 0:
        return res, err
    else:
        return 1.0 / res, err / (res ** 2) + eps / res # Just applied the propagation error formula
```

```python
x = - 1.0 # Compute 1 / e
res_array = np.array([taylor(x, N) for N in range(1, 100, 1)]) # Try different values of N
res_array
```

```python
res = res_array[np.argmin(res_array[:,1])] # Minimum error
print(f"result = {res[0]:.15f} +- {res[1]:.15f}")
print(f"1 / e = {(1 / np.e):.15f}")
print(f"Best N: {np.argmin(res_array[:,1]) + 1}")
```

```python
x = 0.5 # Compute sqrt(e)
res_array = np.array([taylor(x, N) for N in range(1, 100, 1)]) # Different values of N
res_array
```

```python
res = res_array[np.argmin(res_array[:,1])] # Minimum error
print(f"result = {res[0]:.15f} +- {res[1]:.15f}")
print(f"sqrt(e) = {(np.sqrt(np.e)):.15f}")
print(f"Best N: {np.argmin(res_array[:,1]) + 1}")
```

```python
def N_opt(x): # Computes the best values of N
    global eps

    best_N = 1
    while np.abs(x) ** (best_N + 1) / factorial(best_N + 1) > eps:
        best_N += 1

    return best_N
```

When $N$ is sufficient large, the error reaches a minimum. $N_{opt}$ can be computed roughly as:

$$
    \frac{|x| ^ {N_{opt} + 1}}{\left( N_{opt} + 1 \right)!} \approx u
$$

```python
N_opt(- 1.0), N_opt(0.5), N_opt(3.0), N_opt(- 30.0) # Our best Ns estimations
```
