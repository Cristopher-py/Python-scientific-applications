### Floating-Point Errors Table

This summary covers the total error propagation for the four basic arithmetic operations, assuming $a, b > 0$ are machine numbers (floating-point representations).

Let $fl(x)$ denote the floating-point representation of a real number $x$. We assume the standard model:

- $fl(a) = a(1 + \delta_a)$, with $|\delta_a| \le u$ (unit roundoff, e.g., $u \approx 1.1 \times 10^{-16}$ for double precision)
- $fl(b) = b(1 + \delta_b)$, with $|\delta_b| \le u$
- For any basic operation $\circ$ (+, -, *, /), the computed result satisfies:
  $fl(a \circ b) = (a \circ b)(1 + \delta_{\circ})$, with $|\delta_{\circ}| \le u$

| Operation | Computed Value | Relative Error | Absolute Error | Upper Bound Relative Error | Numerical Stability |
|---|---|---|---|---|---|
| **Addition** <br> $a + b$ | $(a(1+\delta_a) + b(1+\delta_b))(1+\delta_{\oplus})$ | $\frac{a}{a+b}\delta_a + \frac{b}{a+b}\delta_b + \delta_{\oplus}$ | $a\delta_a + b\delta_b + (a+b)\delta_{\oplus}$ | $\approx 2u$               | **Stable** (always) |
| **Subtraction** <br> $a - b$ | $(a(1+\delta_a) - b(1+\delta_b))(1+\delta_{\ominus})$ | $\frac{a}{a-b}\delta_a - \frac{b}{a-b}\delta_b + \delta_{\ominus}$ | $a\delta_a - b\delta_b + (a-b)\delta_{\ominus}$ | $\approx \frac{a+b}{\lvert a-b \rvert} u$ | **Unstable** if $a \approx b$ <br> (Catastrophic cancellation) |
| **Multiplication** <br> $a \cdot b$ | $ab(1+\delta_a)(1+\delta_b)(1+\delta_{\odot})$ | $\delta_a + \delta_b + \delta_{\odot}$ | $ab(\delta_a + \delta_b + \delta_{\odot})$ | $\approx 3u$               | **Stable** (always) |
| **Division** <br> $a / b$ | $\frac{a}{b} \cdot \frac{1+\delta_a}{1+\delta_b}(1+\delta_{\oslash})$ | $\delta_a - \delta_b + \delta_{\oslash}$ | $\frac{a}{b}(\delta_a - \delta_b + \delta_{\oslash})$ | $\approx 3u$               | **Stable** (provided $b \neq 0$) |
