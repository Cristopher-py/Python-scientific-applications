### Neumann Series

Given a real square matrix $T$, the Neumann series is an operator defined as:

$$
    N \triangleq \sum_{k = 0}^{\infty} \, T^k
$$

If the spectral radius of $T$ is less than 1, the Neumann series converges to:

$$
    N = \left( I - T \right)^{-1}
$$

Where $I$ is the identity matrix.

The spectral radius of $T$ is defined as the maximum of the absolute values of its eigenvalues:

$$
    \rho(T) \triangleq \text{max} \left\{ |\lambda_1|, |\lambda_2|, |\lambda_3|, \dots, |\lambda_n| \right\}
$$

$N$ is finite if:

$$
    \rho(T) < 1
$$

We can use this to solve linear systems. Consider:

$$
    A x = b
$$

Define:

$$
    H \triangleq I - A
$$

So, the solution can be computed as:

$$
    x = \left( I - H \right)^{-1} b
$$

If $\rho(H) < 1$, we can write:

$$
    \left( I - H \right)^{-1} = \sum_{k = 0}^{\infty} \, H^k
$$

And the solution is:

$$
    x = \sum_{k = 0}^{\infty} \, H^{k} b \qquad \rho(H) < 1
$$

```python
import numpy as np
```

```python
n = 100 # A has dimension 100 x 100

A = np.random.rand(n, n) / n # Random matrix
A = A @ (A.T) + np.eye(n)  # Symmetric positive definite
b = 2.0 * np.random.rand(n) # Random vector

A, b
```

```python
H = np.eye(n) - A # H = I - A
H
```

```python
def neumann_series(A, b, eps = 1e-09, max_iter = 100): # Solves Ax = b applying the Neumann series

    n = A.shape[0] # Dimension of A
    I = np.eye(n) # Identity matrix
    H = I - A # H Matrix

    # Check the spectral radius. It must be less than 1
    eigvals = np.linalg.eigvals(H)
    spectral_radius = np.max(np.abs(eigvals))
    if spectral_radius >= 1.0: raise ValueError(f"Neumann series does not converge: spectral radius = {spectral_radius} >= 1")

    x = np.zeros_like(b)
    term = b.copy()

    for j in range(max_iter): # Computing the Neumann series
        x += term
        
        residual = b - (A @ x) # Computing the residual b - A @ x
        residual_norm = np.linalg.norm(residual) # Computing the residual norm

        if residual_norm < eps: break # If the residual's norm is less than the tolerance threshold value, halt the loop

        term = H @ term # Update the next term

    else:
        print("Maximum number of iterations reached")

    info = { # General info
        "N iterations": j + 1,
        "Residual norm": residual_norm,
        "Spectral radius": spectral_radius,
        "Condition number of A": np.linalg.cond(A),
    }

    return x, info
```

```python
neumann_sol, info = neumann_series(A, b) # Solution by Neumann series
neumann_sol, info
```

```python
numpy_sol = np.linalg.solve(A, b) # Solution by NumPy (direct solver)
numpy_sol
```

```python
np.max(np.abs(numpy_sol - neumann_sol)) # Maximum absolute error found
```

To estimate a posteriori error vector $\varepsilon$ without knowing the true solution, we can solve:

$$
    A \varepsilon = b - A x^{*}
$$

Where $x^{*}$ is the approximated solution yield by our solver

```python
residual = b - A @ neumann_sol # Computing the residual
abs_err = np.abs(neumann_series(A, residual, eps = 1e-13)[0]) # Absolute error vector
rel_err = abs_err / np.abs(neumann_sol) * 100.0 # Relative error vector
abs_err, rel_err
```

Let's test slow convergence

```python
A = np.array([[0.9, 0.4, 0.5],
              [0.3, 0.9, 0.6],
              [0.4, 0.5, 0.9]])
b = np.array([2.0, -1.0, 3.0])

neumann_series(A, b)
```

```python
# Error estimation
neumann_series(A, b - (A @ (neumann_series(A, b)[0])), eps = 1e-13)
```
