### Finite difference absolute error

When we compute numerically the derivative of a function with respect of $x$ using the forward-step formula:

$$
    f'(x) \approx \frac{f(x + h) - f(x)}{h}
$$

We carry an error as well:

$$
    E \lesssim \frac{2 \varepsilon_{mach}}{h} \left| f(x) \right| + \frac{h}{2} \left| f''(x) \right| + \varepsilon_{mach} \left|x \right| \left|f''(x) \right|
$$

- $\frac{2 \varepsilon_{mach}}{h} \left| f(x) \right|$ is the roundoff error from function evaluations;
- $\frac{h}{2} \left| f''(x) \right|$ is the truncation error from the finite-difference approximation;
- $\varepsilon_{mach} \left|x \right| \left|f''(x) \right|$ is the propagated error from the inaccurate representation of $x$;

For a double, $\varepsilon_{mach} \approx 2.2 \times 10^{-16}$

The best value of $h$ to minimize the error is:

$$
    h_{opt} = 2 \, \sqrt{\frac{\left| f(x) \right|}{\left| f''(x) \right|}} \, \sqrt{\varepsilon_{mach}}
$$

Let's see these formulas in action

```python
import numpy as np

eps = 2.2e-16 # Machine precision for doubles
```

```python
f = lambda x: (np.cos(x) + np.sin(x)) * np.exp(x) # Test function: f(x) = (cos(x) + sin(x)) * e^x
f2 = lambda x: - 2 * np.exp(x) * (np.sin(x) - np.cos(x)) # second derivative

x = np.pi / 3.0 # point where to evaluate the first derivative

def derivative(f, f2, x, h):

    global eps

    d = (f(x + h) - f(x)) / h # Computing the derivative
    err = h / 2.0 * np.abs(f2(x)) + 2.0 * eps * np.abs(f(x)) / h + np.abs(f2(x)) * np.abs(x) * eps # Error formula

    return d, err
```

```python
d = []
err = []
h = 10.0 ** (- np.arange(1, 20 + 1)) # Let's test some h steps (1e-01, 1e-02, 1e-03, ... , 1e-20)

for hv in h:
    dv, errv = derivative(f, f2, x, hv)
    d.append(dv)
    err.append(errv)

d, err = np.array(d), np.array(err)
d, err
```

As you can see, if $h$ is chosen too small the derivative is absolutely equal to zero and the error explodes

```python
h_opt = 2 * np.sqrt(eps * np.abs(f(x)) / np.abs(f2(x))) # Best value for h
h_opt
```

```python
derivative(f, f2, x, h_opt) # Best result obtained
```
