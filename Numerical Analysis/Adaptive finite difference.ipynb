```python
import numpy as np
```

```python
def f(x): # Target function
    return x ** 2 / (x ** 3 + 1.0)
```

```python
def dev(x0, h, scheme): # Finite difference derivative
    match scheme:
        case 'forward':
            return (f(x0 + h) - f(x0)) / h
        case 'backward':
            return (f(x0) - f(x0 - h)) / h
        case 'central':
            return (f(x0 + h) - f(x0 - h)) / (2 * h)
        case _:
            return 0.0
```

```python
def dev_adaptive(f, x0, h0 = 1.0, eps = 1e-03, scheme = 'forward'): # Adaptive finite difference

    h = 2.0 * h0
    j = 0
    err = eps + 1.0

    while err > eps: # Absolute error criterion

        h /= 2.0
        j += 1

        derivative = dev(x0, h, scheme)
        derivative2 = dev(x0, h / 2.0, scheme)
        err = np.abs(derivative2 - derivative)

    print(f"Loop ended at j = {j}. h = {h}")
    return derivative, err # It returns the derivative and the absolute error committed
```

```python
x0 = 1.0 # Point target
h0 = 0.1 # Initial step
eps = 1e-06 # Absolute tolerance
```

```python
dev_adaptive( # Forward scheme
    f = f,
    x0 = x0,
    h0 = h0,
    eps = eps,
    scheme = 'forward'
)
```

$$
    f'(1)_{forward} = 0.249998 \pm 0.000001
$$

```python
dev_adaptive( # Backward scheme
    f = f,
    x0 = x0,
    h0 = h0,
    eps = eps,
    scheme = 'backward'
)
```

$$
    f'(1)_{backward} = 0.250002 \pm 0.000001
$$

```python
dev_adaptive( # Central scheme
    f = f,
    x0 = x0,
    h0 = h0,
    eps = eps,
    scheme = 'central'
)
```

$$
    f'(1)_{central} = 0.2500008 \pm 0.0000006
$$
