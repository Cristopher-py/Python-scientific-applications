### Relative errors of products

Given a product of values:

$$
    P = \prod_{j = 1}^{N} \, x_j
$$

We're going to compute both inherited and algorithmic error, to obtain a total error upper bound

```python
# Libraries
from itertools import combinations
import sympy as smp
from math import prod
```

```python
x1, x2, x3, x4, x5, x6, x7, x8 = smp.symbols("x_1:9", real = True) # 8 total variables x_1, x_2, ..., x_8. Not necessary all positive!
eps1, eps2, eps3, eps4, eps5, eps6, eps7, eps8 = smp.symbols("epsilon_1:9", real = True) # RELATIVE errors of variables x_1, x_2, ..., x_8. Not necessary all positive!

x_list = [x1, x2, x3, x4, x5, x6, x7, x8]
eps_list = [eps1, eps2, eps3, eps4, eps5, eps6, eps7, eps8]
```

```python
product = prod(x_list) # Product for N = 8
product
```

The relative inherited error is given by:

$$
    \epsilon_{inh} = \left| \frac{ \prod_{j = 1}^{N} \, x_j \, \frac{\partial P}{\partial x_j} \, \epsilon_j }{P} \right|
$$

```python
sum([smp.Abs(x * smp.diff(product, x, 1) * eps / product) for x, eps in zip(x_list, eps_list)]).together() # Relative inherited error
```

Our relative inherited error is bounded by:

$$
    \epsilon_{inh} \leq 8 u
$$

For $N$ terms:

$$
    \boxed{\epsilon_{inh} \leq N u}
$$

Let's now compute the algorithmic error using the sequential method

```python
def fl(z: smp.Symbol, err_z: smp.Symbol): # float operator
    return z * (1 + err_z)
```

```python
eta1, eta2, eta3, eta4, eta5, eta6, eta7 = smp.symbols("eta_1:8", real = True) # Atomic relative algorithmic errors
eta_list = [eta1, eta2, eta3, eta4, eta5, eta6, eta7]

# Sequencial product: x_1 * x_2 * ... * x_8
op1 = fl(x1 * x2, eta1)
op2 = fl(op1 * x3, eta2)
op3 = fl(op2 * x4, eta3)
op4 = fl(op3 * x5, eta4)
op5 = fl(op4 * x6, eta5)
op6 = fl(op5 * x7, eta6)
op7 = fl(op6 * x8, eta7)

op7 # Final result
```

```python
subs = {a * b: 0 for a, b in combinations(eta_list, 2)} # Linearization dictionay for subs() method

# Computing the relative algorithmic error
err_algo = (op7 - product).expand().subs(subs) / product
err_algo = smp.Abs(err_algo.collect(x_list))
err_algo
```

This quantity has upper bound:

$$
    \epsilon_{algo} \leq 7 u
$$

For a product of $N$ terms:

$$
    \boxed{\epsilon_{algo} \leq (N - 1) \, u}
$$

Then, the upper bound of the total relative error (relative inherited error + relative algorithmic error) for $N$ terms is given by:

$$
    \boxed{\epsilon_{tot} \leq u \, (2 N - 1)}
$$
