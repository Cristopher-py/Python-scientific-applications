Given a square linear system $A \, \vec{x} = \vec{b}$, the error $\vec{\varepsilon}$ of the solution $\vec{X}$, computed numerically, solves the linear system:

$$
    A \, \vec{\varepsilon} = \vec{r}
$$

Where $\vec{r} = \vec{b} - A \, \vec{X}$

```python
import numpy as np
```

```python
def gauss(A, b): # Gauss elimination algorithm from scratch

    # Convert to float and check dimensions
    A = np.asarray(A, dtype = float)
    b = np.asarray(b, dtype = float).flatten()

    n = A.shape[0] # Dimensions of A

    # Dimensions check
    if A.shape != (n, n) or len(b) != n:
        raise ValueError("A must be square and b must have length N.")
    if n == 0:
        return np.array([])

    U = np.column_stack((A, b)) # Build the augmented matrix A | b

    # Forward elimination with partial pivoting
    for i in range(n):
        pivot = np.argmax(np.abs(U[i:, i])) + i
        if np.abs(U[pivot, i]) < 1e-12: # Instability detected
            raise np.linalg.LinAlgError("Matrix is singular.")

        U[[i, pivot]] = U[[pivot, i]] # Swap pivot row (does nothing if pivot == i)

        # Vectorized elimination: update all rows below 'i' at once
        factors = U[i + 1:, i] / U[i, i]
        U[i + 1:, i:] -= factors[:, np.newaxis] * U[i, i:]

    # Back substitution
    X = np.zeros(n)
    for i in range(n - 1, - 1, - 1):
        X[i] = (U[i, - 1] - np.dot(U[i, i + 1: n], X[i + 1:])) / U[i, i]

    return X
```

### Testing a random linear system

```python
N = 100 # A has dimension 100 x 100

A = np.random.rand(N, N) # Random matrix
A = A @ (A.T) + np.eye(N)  # Symmetric positive definite
b = np.random.rand(N) # Random vector

A, b
```

```python
X = gauss(A, b) # Solution
X
```

```python
np.linalg.solve(A, b) # Solution X from NumPy perfectly matches our solution
```

```python
res = b - A @ X # Residual vector
res
```

```python
errors = np.abs(gauss(A, res)) # Error estimation (taking absolute values)
errors
```

```python
np.min(errors), np.max(errors) # Minimum and maximum estimated absolute error
```

```python
relative_errors = errors / np.abs(X) * 1e02 # Estimated relative errors [%]
relative_errors
```

```python
np.min(relative_errors), np.max(relative_errors) # Minimum and maximum relative errors [%]
```

### Testing the inverse matrix

```python
A = 2.0 * np.eye(3) + np.eye(3, k = 1) + np.eye(3, k = - 1) # Matrix target
A
```

```python
# Known vectors
e1, e2, e3 = (np.array([1.,
                       0.,
                       0.]),
              np.array([0.,
                        1.,
                        0.]),
              np.array([0.,
                        0.,
                        1.]),)

e1, e2, e3
```

```python
row1, row2, row3 = gauss(A, e1), gauss(A, e2), gauss(A, e3) # A^{-1}
row1, row2, row3
```

```python
np.linalg.inv(A) # Inverse of A using NumPy
```

```python
# Computing absolute errors
res1, res2, res3 = e1 - A @ row1, e2 - A @ row2, e3 - A @ row3
err1, err2, err3 = np.abs(gauss(A, res1)), np.abs(gauss(A, res2)), np.abs(gauss(A, res3))

err1, err2, err3
```

```python
err1 / np.abs(row1) * 1e02, err2 / np.abs(row2) * 1e02, err3 / np.abs(row3) * 1e02 # Relative errors [%]
```
