```python
# Importing basic staff
from sympy import symbols, Function, Matrix, cos, sin, diff

# Dot notation for derivatives with respect time
from sympy.physics.vector import init_vprinting
init_vprinting()
```

```python
# time and functions(t)
t = symbols('t')

rho = Function('rho', real = True, nonnegative = True)(t)
theta = Function('theta', real = True)(t)
phi = Function('phi', real = True)(t)
```

In polar coordinates the orthonormal basis vectors are defined as:

$$
    \hat{\rho} \triangleq \begin{pmatrix} \cos\phi \\ \sin\phi \end{pmatrix}, \qquad
    \hat{\phi} \triangleq \begin{pmatrix} -\sin\phi \\ \cos\phi \end{pmatrix}
$$

In Cartesian coordinates

```python
# Position of a point in polar coordinate
r_polar = Matrix([rho * cos(phi),
                  rho * sin(phi)])
r_polar
```

$$
    r = \begin{pmatrix} \rho \cos\phi \\ \rho \sin\phi \end{pmatrix}_{\text{cartesian}} = \begin{pmatrix} \rho \\ 0 \end{pmatrix}_{\text{polar}}
$$

```python
# Velocity in polar coordinates
v_polar = diff(r_polar, t, 1)
v_polar
```

$$
    v = \begin{pmatrix} -\rho \sin\phi \,\dot{\phi} + \cos\phi \,\dot{\rho} \\ \rho \cos\phi \,\dot{\phi} + \sin\phi \,\dot{\rho} \end{pmatrix}_{\text{cartesian}} = \begin{pmatrix} \dot{\rho} \\ \rho \dot{\phi} \end{pmatrix}_{\text{polar}}
$$

```python
# Acceleration in polar coordinates
a_polar = diff(v_polar, t, 1)
a_polar
```

$$
    a = \begin{pmatrix}
        - \rho \sin\phi \, \ddot{\phi} - \rho \cos\phi \, \dot{\phi}^{2} - 2 \sin\phi \, \dot{\phi} \dot{\rho} + \cos\phi \, \ddot{\rho} \\
        - \rho \sin\phi \, \dot{\phi}^{2} + \rho \cos\phi \, \ddot{\phi} + \sin\phi \, \ddot{\rho} + 2 \cos\phi \, \dot{\phi} \dot{\rho}
    \end{pmatrix}_{cartesian}
    = \begin{pmatrix}
        \ddot{\rho} - \rho \dot{\phi}^{2} \\
        \rho \ddot{\phi} + 2 \dot{\rho} \dot{\phi}
    \end{pmatrix}_{polar}
$$

---

In spherical coordinates (using $\rho$ for the radial coordinate, $\theta$ for the polar angle from the $z$-axis, and $\phi$ for the azimuthal angle) the orthonormal basis vectors are defined as:

$$
    \hat{\rho} \triangleq \begin{pmatrix} \sin\theta \cos\phi \\ \sin\theta \sin\phi \\ \cos\theta \end{pmatrix}, \qquad
    \hat{\theta} \triangleq \begin{pmatrix} \cos\theta \cos\phi \\ \cos\theta \sin\phi \\ -\sin\theta \end{pmatrix}, \qquad
    \hat{\phi} \triangleq \begin{pmatrix} -\sin\phi \\ \cos\phi \\ 0 \end{pmatrix}
$$

```python
# Position of a point in spherical coordinates
r_spherical = Matrix([rho * cos(phi) * sin(theta),
                      rho * sin(phi) * sin(theta),
                      rho * cos(theta)])
r_spherical
```

$$
    r = \begin{pmatrix} \rho \sin\theta \cos\phi \\ \rho \sin\theta \sin\phi \\ \rho \cos\theta \end{pmatrix}_{cartesian}
    = \begin{pmatrix} \rho \\ 0 \\ 0 \end{pmatrix}_{spherical}
$$

```python
# Velocity in spherical coordinates
v_spherical = diff(r_spherical, t, 1)
v_spherical
```

$$
    v = \begin{pmatrix} 
        - \rho \sin\phi \sin\theta \,\dot{\phi} + \rho \cos\phi \cos\theta \,\dot{\theta} + \sin\theta \cos\phi \,\dot{\rho} \\ 
        \rho \sin\phi \cos\theta \,\dot{\theta} + \rho \sin\theta \cos\phi \,\dot{\phi} + \sin\phi \sin\theta \,\dot{\rho} \\ 
        - \rho \sin\theta \,\dot{\theta} + \cos\theta \,\dot{\rho} 
    \end{pmatrix}_{cartesian}
    = \begin{pmatrix} 
        \dot{\rho} \\ 
        \rho \dot{\theta} \\ 
        \rho \sin\theta \,\dot{\phi} 
    \end{pmatrix}_{spherical}
$$

```python
# Acceleration of a point in spherical coordinates
a_spherical = diff(v_spherical, t, 1)
a_spherical
```

$$
    a = \begin{pmatrix}
        - \rho \sin\phi \sin\theta \,\ddot{\phi} - 2 \rho \sin\phi \cos\theta \,\dot{\phi} \dot{\theta} - \rho \sin\theta \cos\phi \,\dot{\phi}^{2} - \rho \sin\theta \cos\phi \,\dot{\theta}^{2} + \rho \cos\phi \cos\theta \,\ddot{\theta} - 2 \sin\phi \sin\theta \,\dot{\phi} \dot{\rho} + \sin\theta \cos\phi \,\ddot{\rho} + 2 \cos\phi \cos\theta \,\dot{\rho} \dot{\theta} \\
        - \rho \sin\phi \sin\theta \,\dot{\phi}^{2} - \rho \sin\phi \sin\theta \,\dot{\theta}^{2} + \rho \sin\phi \cos\theta \,\ddot{\theta} + \rho \sin\theta \cos\phi \,\ddot{\phi} + 2 \rho \cos\phi \cos\theta \,\dot{\phi} \dot{\theta} + \sin\phi \sin\theta \,\ddot{\rho} + 2 \sin\phi \cos\theta \,\dot{\rho} \dot{\theta} + 2 \sin\theta \cos\phi \,\dot{\phi} \dot{\rho} \\
        - \rho \sin\theta \,\ddot{\theta} - \rho \cos\theta \,\dot{\theta}^{2} - 2 \sin\theta \,\dot{\rho} \dot{\theta} + \cos\theta \,\ddot{\rho}
    \end{pmatrix}_{cartesian}
    = \begin{pmatrix}
        \ddot{\rho} - \rho \dot{\theta}^2 - \rho \sin^2\theta \,\dot{\phi}^2 \\
        \rho \ddot{\theta} + 2\dot{\rho}\dot{\theta} - \rho \sin\theta \cos\theta \,\dot{\phi}^2 \\
        \rho \sin\theta \,\ddot{\phi} + 2\dot{\rho}\sin\theta \,\dot{\phi} + 2\rho \cos\theta \,\dot{\theta}\dot{\phi}
    \end{pmatrix}_{spherical}
$$
