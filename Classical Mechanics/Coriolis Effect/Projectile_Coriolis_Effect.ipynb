## Projectile motion with Coriolis force

The equations of motion for a projectile in a local East-North-Up (ENU) frame rotating with the Earth, including the Coriolis force but neglecting air resistance and centrifugal effects (absorbed into effective gravity), are:

$$
    \frac{d^2\mathbf{r}}{dt^2} = \mathbf{g} - 2\,\boldsymbol{\Omega} \times \frac{d\mathbf{r}}{dt}
$$

where:

$r = (x, y, z)$ is the position vector, \
$g = (0, 0, - g)$ is the effective gravitational acceleration (downward), \
$\Omega = (0,\, \Omega\cos\phi,\, \Omega\sin\phi)$ is the Earth’s angular velocity vector in the ENU frame, \
$\phi$ = is the latitude.

Expanding in components we obtain:

\begin{cases}
    \ddot{x} = 2\Omega\sin\phi\,\dot{y} - 2\Omega\cos\phi\,\dot{z} \\
    \ddot{y} = -2\Omega\sin\phi\,\dot{x} \\
    \ddot{z} = -g + 2\Omega\cos\phi\,\dot{x}
\end{cases}

This system admits an exact analytical solution without any assumptions or simplification. The equations are linear, constant-coefficient ODEs and can be solved in closed form using elementary functions. Maxima CAS gives the following solution (code below):
