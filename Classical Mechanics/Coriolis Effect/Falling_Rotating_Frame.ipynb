## Falling object in a rotating frame full solution

The equations of motion for a dropped object from height $h$, studied in a local East-North-Up (ENU) frame rotating with the Earth, including the Coriolis and centrifugal forces but neglecting air resistance, are:

$$
    \ddot{\mathbf{r}} = \mathbf{g} - 2\,\boldsymbol{\Omega} \times \dot{\mathbf{r}} - \boldsymbol{\Omega} \times (\boldsymbol{\Omega} \times \mathbf{r})
$$

where:

$r = (x, y, z)$ is the position vector, \
$g = (0, 0, - g)$ is the effective gravitational acceleration (downward), \
$\Omega = (0,\, \Omega\cos\phi,\, \Omega\sin\phi)$ is the Earth’s angular velocity vector in the ENU frame, \
$\phi$ = is the latitude.

Expanding in components we obtain:

\begin{cases}
    \ddot{x} = 2\Omega \sin\varphi \, \dot{y} - 2\Omega \cos\varphi \, \dot{z} + \Omega^2 x \\
    \ddot{y} = -2\Omega \sin\varphi \, \dot{x} + \Omega^2 \sin^2\varphi \, y - \Omega^2 \sin\varphi \cos\varphi \, z \\
    \ddot{z} = -g + 2\Omega \cos\varphi \, \dot{x} + \Omega^2 \cos^2\varphi \, z - \Omega^2 \sin\varphi \cos\varphi \, y
\end{cases}

Initial conditions are:

\begin{cases}
    x(0) = 0 \\
    y(0) = 0 \\
    z(0) = h \\
    \dot{x}(0) = 0 \\
    \dot{y}(0) = 0 \\
    \dot{z}(0) = 0 \\
\end{cases}

(Generally speaking, the differential equations system above describes the projectile motion in the ENU local frame).

This system admits an exact analytical solution without any assumptions or simplification. Maxima CAS gives the following solution (code below):

```maxima
(%i5)	/* Declare and assume*/
	declare(g, constant)$
	declare(Ω, constant)$
	declare(φ, constant)$
	declare(h, constant)$
	assume(φ >= - %pi / 2, φ <= %pi / 2, g > 0, Ω > 0, t >= 0, h > 0)$
(%i6)	/* First equation*/
(eq1)	-(2*Ω*cos(φ)*('diff(z(t),t)))+2*Ω*sin(φ)*('diff(y(t),t))+Ω^2*x(t)='diff(x(t),t,2)
(%i7)	/* Second equation*/
	eq2: - 2 * Ω * sin(φ) * diff(x(t), t, 1) + Ω ^ 2 * (sin(φ)) ^ 2 * y(t) - Ω ^ 2 * sin(φ) * cos(φ) * z(t) = diff(y(t), t, 2);
(eq2)	-(2*Ω*sin(φ)*('diff(x(t),t)))-Ω^2*cos(φ)*sin(φ)*z(t)+Ω^2*sin(φ)^2*y(t)=
'diff(y(t),t,2)
(%i8)	/* Third equation*/
	eq3: - g + 2 * Ω * cos(φ) * diff(x(t), t, 1) + Ω ^ 2 * (cos(φ)) ^ 2 * z(t) - Ω ^ 2 *  sin(φ) * cos(φ) * y(t) = diff(z(t), t, 2);
(eq3)	2*Ω*cos(φ)*('diff(x(t),t))+Ω^2*cos(φ)^2*z(t)-Ω^2*cos(φ)*sin(φ)*y(t)-g=
'diff(z(t),t,2)
(%i14)	/* Values at t = 0*/
	atvalue(x(t), t=0, 0)$
	atvalue(y(t), t=0, 0)$
	atvalue(z(t), t=0, h)$
	atvalue(diff(x(t), t, 1), t=0, 0)$
	atvalue(diff(y(t), t, 1), t=0, 0)$
	atvalue(diff(z(t), t, 1), t=0, 0)$
(%i16)	/* Declare s complex*/
	declare(s, complex)$
	assume(realpart(s) > 0)$
(%i19)	/* Apply Laplace transform to equations*/
	eq1L: laplace(eq1, t, s);
	eq2L: laplace(eq2, t, s);
	eq3L: laplace(eq3, t, s);
(eq1L)	-(2*Ω*cos(φ)*(s*laplace(z(t),t,s)-h))+2*Ω*sin(φ)*s*laplace(y(t),t,s)+Ω^2*laplace(x(t),t,s)=s^2*laplace(x(t),t,s)
(eq2L)	-(Ω^2*cos(φ)*sin(φ)*laplace(z(t),t,s))+Ω^2*sin(φ)^2*laplace(y(t),t,s)-
2*Ω*sin(φ)*s*laplace(x(t),t,s)=s^2*laplace(y(t),t,s)
(eq3L)	Ω^2*cos(φ)^2*laplace(z(t),t,s)-Ω^2*cos(φ)*sin(φ)*laplace(y(t),t,s)+2*
Ω*cos(φ)*s*laplace(x(t),t,s)-g/s=s^2*laplace(z(t),t,s)-h*s
(%i22)	 /* Simplify notation*/
	eq1L: subst([
	        laplace(x(t), t, s) = X(s),
	        laplace(y(t), t, s) = Y(s),
	        laplace(z(t), t, s) = Z(s)
	    ], eq1L);
	eq2L: subst([
	        laplace(x(t), t, s) = X(s),
	        laplace(y(t), t, s) = Y(s),
	        laplace(z(t), t, s) = Z(s)
	    ], eq2L);
	eq3L: subst([
	        laplace(x(t), t, s) = X(s),
	        laplace(y(t), t, s) = Y(s),
	        laplace(z(t), t, s) = Z(s)
	    ], eq3L);
(eq1L)	-(2*Ω*cos(φ)*(s*Z(s)-h))+2*Ω*sin(φ)*s*Y(s)+Ω^2*X(s)=s^2*X(s)
(eq2L)	-(Ω^2*cos(φ)*sin(φ)*Z(s))+Ω^2*sin(φ)^2*Y(s)-2*Ω*sin(φ)*s*X(s)=s^2*
Y(s)
(eq3L)	Ω^2*cos(φ)^2*Z(s)-Ω^2*cos(φ)*sin(φ)*Y(s)+2*Ω*cos(φ)*s*X(s)-g/s=s^2*
Z(s)-h*s
(%i23)	/* Find X(s), Y(s) and Z(s)*/
	solL: factor(trigsimp(solve([eq1L, eq2L, eq3L], [X(s), Y(s), Z(s)])[1]));
(solL)	[X(s)=-((2*Ω*(h*Ω^2-g)*cos(φ))/(s^2+Ω^2)^2),Y(s)=-
((Ω^2*cos(φ)*sin(φ)*(h*s^4-h*Ω^2*s^2+3*g*s^2+g*Ω^2))/(s^3*(s^2+Ω^2)^2)),Z(s)=(h*s^6+h*Ω^2*
cos(φ)^2*s^4+2*h*Ω^2*s^4-g*s^4-h*Ω^4*cos(φ)^2*s^2+3*g*Ω^2*cos(φ)^2*s^2+h
*Ω^4*s^2-2*g*Ω^2*s^2+g*Ω^4*cos(φ)^2-g*Ω^4)/(s^3*(s^2+Ω^2)^2)]
(%i24)	/* Find x(t) by inverse Laplace transform*/
(%o24)	-(((h*Ω^2-g)*cos(φ)*(sin(Ω*t)-Ω*t*cos(Ω*t)))/Ω^2)
(%i25)	/* Find y(t) by inverse Laplace transform*/
	factor(trigsimp(ilt(rhs(solL[2]), s, t)));
(%o25)	-((cos(φ)*sin(φ)*(2*h*Ω^3*t*sin(Ω*t)-2*g*Ω*t*sin(Ω*t)+2*h*Ω^2*
cos(Ω*t)-2*g*cos(Ω*t)+g*Ω^2*t^2-2*h*Ω^2+2*g))/(2*Ω^2))
(%i26)	/* Find z(t) by inverse Laplace transform*/
	factor(trigsimp(ilt(rhs(solL[3]), s, t)));
(%o26)	(2*h*Ω^3*cos(φ)^2*t*sin(Ω*t)-2*g*Ω*cos(φ)^2*t*sin(Ω*t)+2*h*Ω^2*
cos(φ)^2*cos(Ω*t)-2*g*cos(φ)^2*cos(Ω*t)+g*Ω^2*cos(φ)^2*t^2-g*Ω^2*t^2-
2*h*Ω^2*cos(φ)^2+2*g*cos(φ)^2+2*h*Ω^2)/(2*Ω^2)
```

Maxima CAS returns:

$$
	\begin{aligned}
		x(t) &= \frac{(g - h\Omega^2)\cos\varphi}{\Omega^2}\Bigl(\sin(\Omega t) - \Omega t\cos(\Omega t)\Bigr) \\
		y(t) &= \frac{\cos\varphi\sin\varphi}{\Omega^2}(g - h\Omega^2)\Bigl(\Omega t\sin(\Omega t) + \cos(\Omega t) - 1\Bigr) - \frac{g}{2}\,\cos\varphi\sin\varphi\; t^2\\
		z(t) &= \frac{1}{2\Omega^2}\Bigl[2\cos^2\varphi\,(h\Omega^2 - g)\bigl(\Omega t\sin(\Omega t)+\cos(\Omega t)-1\bigr) - g\Omega^2 t^2\sin^2\varphi + 2g\cos^2\varphi + 2h\Omega^2\sin^2\varphi\Bigr]
	\end{aligned}
$$

By Applying Taylor series expansion at $\Omega$ near zero, we get:

```maxima
(%i27)	/* Find x(t) by inverse Laplace transform*/
	xt: factor(trigsimp(ilt(rhs(solL[1]), s, t)));
(xt)	-(((h*Ω^2-g)*cos(φ)*(sin(Ω*t)-Ω*t*cos(Ω*t)))/Ω^2)
(%i28)	/* Find y(t) by inverse Laplace transform*/
	yt: factor(trigsimp(ilt(rhs(solL[2]), s, t)));
(yt)	-((cos(φ)*sin(φ)*(2*h*Ω^3*t*sin(Ω*t)-2*g*Ω*t*sin(Ω*t)+2*h*Ω^2*cos(Ω*t)-2*g*cos(Ω*t)+g*Ω^2*t^2-2*h*Ω^2+2*g))/(2*Ω^2))
(%i29)	/* Find z(t) by inverse Laplace transform*/
	zt: factor(trigsimp(ilt(rhs(solL[3]), s, t)));
(zt)	(2*h*Ω^3*cos(φ)^2*t*sin(Ω*t)-2*g*Ω*cos(φ)^2*t*sin(Ω*t)+2*h*Ω^2*cos(φ)^2*
cos(Ω*t)-2*g*cos(φ)^2*cos(Ω*t)+g*Ω^2*cos(φ)^2*t^2-g*Ω^2*t^2-2*h*Ω^2*
cos(φ)^2+2*g*cos(φ)^2+2*h*Ω^2)/(2*Ω^2)
(%i46)	taylor(xt, Ω, 0, 3);
(%o46)/T/	(g*cos(φ)*t^3*Ω)/3-((g*cos(φ)*t^5+10*h*cos(φ)*t^3)*Ω^3)/30+...
(%i47)	taylor(yt, Ω, 0, 4);
(%o47)/T/	-(((g*cos(φ)*sin(φ)*t^4+4*h*cos(φ)*sin(φ)*t^2)*Ω^2)/8)+
((g*cos(φ)*sin(φ)*t^6+18*h*cos(φ)*sin(φ)*t^4)*Ω^4)/144+...
(%i48)	taylor(zt, Ω, 0, 2);
(%o48)/T/	-((g*t^2-2*h)/2)+((g*cos(φ)^2*t^4+4*h*cos(φ)^2*t^2)*Ω^2)/8+...
```

$$
	\begin{aligned}
		x(t) &= \frac{g\cos\varphi}{3}\, t^3\,\Omega \;-\; \frac{\cos\varphi}{30}\bigl(g\,t^5 + 10h\,t^3\bigr)\,\Omega^3 \;+\; \cdots \\
		y(t) &= -\frac{\cos\varphi\sin\varphi}{8}\bigl(g\,t^4 + 4h\,t^2\bigr)\,\Omega^2 \;+\; \frac{\cos\varphi\sin\varphi}{144}\bigl(g\,t^6 + 18h\,t^4\bigr)\,\Omega^4 \;+\; \cdots \\
		z(t) &= h - \frac{g}{2}\,t^2 \;+\; \frac{\cos^2\varphi}{8}\bigl(g\,t^4 + 4h\,t^2\bigr)\,\Omega^2 \;+\; \cdots
	\end{aligned}
$$

Let's compute the eastward and northward deflections. Just substitute $t = \sqrt{2 h / g}$ into $x(t)$ and $y(t)$:

```maxima
(%i51)	D_east: factor(subst(t = sqrt(2 * h / g), xt));
(D_east)	-(((h*Ω^2-g)*(sqrt(g)*sin((sqrt(2)*sqrt(h)*Ω)/sqrt(g))-sqrt(2)*sqrt(h)*Ω*cos((sqrt(2)*sqrt(h)*Ω)/sqrt(g)))*cos(φ))/(sqrt(g)*Ω^2))
(%i53)	taylor(D_east, Ω, 0, 3);
(%o53)/T/	(sqrt(2)^3*sqrt(h)^3*cos(φ)*Ω)/(3*sqrt(g))-(sqrt(2)^5*sqrt(h)^5*cos(φ)*Ω^3)/(5*sqrt(g)^3)+...
(%i54)	D_north: factor(subst(t = sqrt(2 * h / g), yt));
(D_north)	-(((sqrt(2)*h^(3/2)*Ω^3*sin((sqrt(2)*sqrt(h)*Ω)/sqrt(g))-sqrt(2)*g*sqrt(h)*Ω*sin((sqrt(2)*sqrt(h)*Ω)/sqrt(g))+sqrt(g)*h*Ω^2*
cos((sqrt(2)*sqrt(h)*Ω)/sqrt(g))-g^(3/2)*cos((sqrt(2)*sqrt(h)*Ω)/sqrt(g))+g^(3/2))*cos(φ)*sin(φ))/(sqrt(g)*Ω^2))
(%i56)	taylor(D_north, Ω, 0, 4);
(%o56)/T/	-((3*h^2*cos(φ)*sin(φ)*Ω^2)/(2*g))+(5*h^3*cos(φ)*sin(φ)*Ω^4)/(9*g^2)+...
```

Results:

$$
	\begin{aligned}
		D_{\text{east}} &= \frac{(g - h\Omega^2)\cos\varphi}{\Omega^2}\bigl(\sin\alpha - \alpha\cos\alpha\bigr) \approx \frac{2\sqrt{2}\,h^{3/2}\cos\varphi}{3\sqrt{g}}\,\Omega
		\;-\; \frac{4\sqrt{2}\,h^{5/2}\cos\varphi}{5g^{3/2}}\,\Omega^3 \;+\; \cdots \\
		
		D_{\text{north}} &= \frac{\cos\varphi\sin\varphi}{\Omega^2}\Bigl((g - h\Omega^2)(\alpha\sin\alpha + \cos\alpha) - g\Bigr) \approx -\frac{3h^2\cos\varphi\sin\varphi}{2g}\,\Omega^2
		\;+\; \frac{5h^3\cos\varphi\sin\varphi}{9g^2}\,\Omega^4 \;+\; \cdots
	\end{aligned}
$$

where $\displaystyle \alpha = \Omega\sqrt{\frac{2h}{g}}$.
